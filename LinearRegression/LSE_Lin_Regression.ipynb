```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#LSE Algorithms
```

```python
df = pd.read_csv('ecommerce.csv')
```

```python
df
```

```python
#Exloratory Data Analysis
```

```python
df.rename(columns = {'Avg. Session Length' : 'ASL', 'Time on App': 'ToA', 'Length of Membership' : 'LoM', 'Yearly Amount Spent':'YAS'}, inplace = True)
```

```python
df1 = df
```

```python
df1['interceft'] = [1] * 500
```

```python
X1 = df1[['interceft','ASL', 'ToA', 'Time on Website', 'LoM']]
```

```python
X1
```

```python
y1 = df1['YAS']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, x_test, y_train, y_test = train_test_split(X1, y1, test_size = 0.3, random_state = 0)
```

```python
Theta = np.linalg.inv((X_train.T) @ X_train)@(X_train.T)@y_train
```

```python
Theta
```

```python
Theta = np.array(Theta)
```

```python
def predict(X_test):
    return X_test @ Theta
```

```python
y_pred = predict(x_test)
```

```python
y_pred
```

```python
plt.scatter(y_pred, y_test, color = 'pink')
plt.xlabel('Prediction')
plt.show()
```

```python
from sklearn.metrics import mean_squared_error as MSE
from sklearn.metrics import mean_absolute_error as MAE
```

```python
print("Mean Square Error: ", MSE(y_pred, y_test))
```

```python
data['b'] = [1] * 500
```

```python
X = np.array(data)
```

```python
X
```

```python
y = np.array(df2['Yearly Amount Spent']).reshape(-1, 1)
```

```python
y
```

```python
w = np.random.rand(5,1)
```

```python
w
```

```python
def grad_descent(w, X, y, L):
    n = len(X)
    error = y - X @ w
    grad = (2/n) * (X.T) @ error
    return w - L*grad
    
```

```python
erochs = 10000
for i in range(erochs):
    if i % 1000 == 0:
        print(f'epoch {i}')
    w = grad_descent(w, X, y, 0.01)
print(w)
```
