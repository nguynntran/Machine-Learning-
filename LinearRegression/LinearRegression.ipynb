```python
def loss_function(a, b, data):
    total_error = 0
    for i in range(len(data)):
        x = data['YearsExperience'][i]
        y = data['Salary'][i]
        total_error += (y - (a * x + b)) ** 2
        return total_error / float(len(data))
    
```

```python
def grad_descent_func(a, b, data, L):
    n = len(data)
    a_grad = 0
    b_grad = 0
    for i in range(len(data)):
        x = data['YearsExperience'][i]
        y = data['Salary'][i]
        a_grad += 2/n * x * ((a * x + b) - y)
        b_grad += 2/n * ((a * x + b) - y)
    return a - L * a_grad, b - L * b_grad
```

```python
erochs = 1000000
a = 0
b = 0
for i in range(erochs):
    if i % 1000 == 0:
        print(f'epoch {i}')
    a, b = grad_descent_func(a, b, data, 0.0001)
print(a,b)
```
