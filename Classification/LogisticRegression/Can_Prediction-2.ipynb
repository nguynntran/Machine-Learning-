```python
def sigmoid(z):
  return 1/(1 + np.exp(-z))
```

```python
def loss_function(X_, y_, theta):
  m = y_.size
  z = (X_) @ theta
  g = sigmoid(z)
  return (y_ @ np.log(g) + (1 - y_) @ np.log(1 - g))/m
```

```python
def gradient_cal(X_, y_, theta):
  m = y_.size
  z = (X_) @ theta
  g = sigmoid(z)
  return ((X_.T) @ (g - y_))/m
```

```python
def gradient_descent(X_, y_, alpha, num_iters):
  n = X_.shape[1]
  theta = np.zeros(n)
  for i in range(num_iters):
    grad = gradient_cal(X_, y_, theta)
    if i % 1000 == 0:
      print(loss_function(X_,y_,theta))
    theta -= alpha * grad
  return theta
```

```python
def predict (X_, theta):
  z = X_ @ theta
  g = sigmoid(z)
  y_pred = np.zeros(len(g))
  for i in range(len(g)):
    if g[i] >= 0.5:
      y_pred[i] = 1
    else:
      y_pred[i] = 0
  return y_pred
```

```python
alpha = 0.001
num_iters = 11000

theta = gradient_descent(X1_train, y1_train, alpha, num_iters)
print(theta)
```

This indicates numerical instability where the sigmoid output `g` is reaching exactly 0 or 1, causing `np.log(0)` errors. To address this, I will add a small epsilon to clip the sigmoid output `g` values within the `loss_function` and `gradient_cal` functions, preventing them from becoming exactly 0 or 1 before taking the logarithm.



```python
def sigmoid(z):
  return 1/(1 + np.exp(-z))

def loss_function(X_, y_, theta):
  m = y_.size
  z = (X_) @ theta
  g = sigmoid(z)
  epsilon = 1e-10 # Small value to prevent log(0)
  g = np.clip(g, epsilon, 1 - epsilon)
  return -(y_ @ np.log(g) + (1 - y_) @ np.log(1 - g))/m
```

```python
def gradient_cal(X_, y_, theta):
  m = y_.size
  z = (X_) @ theta
  g = sigmoid(z)
  epsilon = 1e-10 # Small value to prevent log(0)
  g = np.clip(g, epsilon, 1 - epsilon)
  return ((X_.T) @ (g - y_))/m
```
